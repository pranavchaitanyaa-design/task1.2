```python
# Dataset 3 - Weight and Height

import pandas as pd
import numpy as np

weight=[55,60,65,70,75,80,68]
height=[150,155,160,165,170,175,168]

print("Mean:",np.mean(weight))
print("Median:",np.median(weight))
print("varience:",np.var(weight))
print("standed deviation:",np.std(weight))
print("Correlation:",np.corrcoef(weight,height))

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=height, y=weight)

plt.title('Weight vs. Height')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 6))
sns.histplot(weight, kde=True)

plt.title('Distribution of Weight')
plt.xlabel('Weight')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
