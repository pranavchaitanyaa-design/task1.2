```python
# Dataset 2 - Product Price and Sales

import pandas as pd
import numpy as np

price=[100,150,200,250,300,180,220]
sales=[500,450,400,350,300,420,380]

print("Mean:",np.mean(price))
print("Median:",np.median(price))
print("varience:",np.var(price))
print("standed deviation:",np.std(price))
print("Correlation:",np.corrcoef(price,sales))

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=sales, y=price)

plt.title('Price vs. Sales')
plt.xlabel('Sales')
plt.ylabel('Price')
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 6))
sns.histplot(price, kde=True)

plt.title('Distribution of Price')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
