```python
import pandas as pd

df = pd.read_csv('/content/gym_members_exercise_tracking_synthetic_data.csv')
print("------Fitness Center Data Analysis--------")
print("Data is loaded  successfully.")

display(df.head())
```

```python
import matplotlib.pyplot as plt

#average calories
avg_calories=df["Calories_Burned"].mean()
print("Average Calories Burned:",avg_calories)

#maximum calories
max_calories=df["Calories_Burned"].max()
print("Maximum Calories Burned:",max_calories)

#minimum calories
min_calories=df["Calories_Burned"].min()
print("Minimum Calories Burned:",min_calories)

#median calories
median_calories=df["Calories_Burned"].median()
print("Median Calories Burned:",median_calories)

#standard deviation
std_calories=df["Calories_Burned"].std()
print("Standard Deviation Calories Burned:",std_calories)

#ploting
plt.figure(figsize=(10,6))
plt.bar(df["Age"],df["Calories_Burned"])
plt.xlabel("Member Age")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis")
plt.show()

plt.figure(figsize=(10,6))
plt.scatter(df["Session_Duration (hours)"], df["Calories_Burned"])
plt.xlabel("Session Duration (hours)")
plt.ylabel("Calories Burned")
plt.title("Calories Burned vs Session Duration")
plt.show()
```
