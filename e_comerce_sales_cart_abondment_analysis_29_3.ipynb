```python
import pandas as pd
import numpy as np
data=pd.DataFrame(
    {
    "customer_name":["sai","pranav","vivek","vijay"],
    "internet_speed":[2,5,3,1],
    "payment_failure":[4,2,3,6],
    "delivery_charges":[121,230,100,216],
    "purchage_completed":[1,0,1,1]
    }
)
data
print("total_abandnment:",(data["purchage_completed"].value_counts()))
print("internet_diagnosis:",(data["internet_speed"].values>=1))
print("payment_diagnosis:",(data["purchage_completed"].values>=1))
print("delivery_diagnosis:",(data["delivery_charges"].value_counts()))
#root cause
print("poor_internet",data["internet_speed"].value_counts())
print("poor_payment",data["payment_failure"].value_counts())
print("poor_delivery",data["delivery_charges"].value_counts())
```

```python
import matplotlib.pyplot as plt
plt.bar(data["customer_name"],data["internet_speed"])
plt.bar(data["customer_name"],data["payment_failure"])
plt.bar(data["customer_name"],data["delivery_charges"])
plt.xlabel("customer_name")
plt.ylabel("diagnosis")
plt.title("diagnosis")
plt.show()
```
