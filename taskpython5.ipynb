```python
# Dataset 4 - Mobile Usage and Battery Drain

import pandas as pd
import numpy as np

usage_hours=[2,4,5,6,7,3,8]
battery_drain=[10,20,28,35,40,15,45]

print("Mean:",np.mean(usage_hours))
print("Median:",np.median(usage_hours))
print("varience:",np.var(usage_hours))
print("standed deviation:",np.std(usage_hours))
print("Correlation:",np.corrcoef(usage_hours,battery_drain))

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=battery_drain, y=usage_hours)

plt.title('Usage Hours vs. Battery Drain')
plt.xlabel('Battery Drain')
plt.ylabel('Usage Hours')
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 6))
sns.histplot(usage_hours, kde=True)

plt.title('Distribution of Usage Hours')
plt.xlabel('Usage Hours')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
