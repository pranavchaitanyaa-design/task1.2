```python
import pandas as pd
import numpy as np

temperature=[32,28,35,30,27,33,31]
icecream_sales=[120,95,150,130,90,145,125]

print("Mean:",np.mean(temperature))
print("Median:",np.median(temperature))
print("varience:",np.var(temperature))
print("standed deviation:",np.std(temperature))
print("Correlation:",np.corrcoef(temperature,icecream_sales))

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=icecream_sales, y=temperature)

plt.title('Temperature vs. Ice Cream Sales')
plt.xlabel('Ice Cream Sales')
plt.ylabel('Temperature')
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 6))
sns.histplot(temperature, kde=True)

plt.title('Distribution of Temperature')
plt.xlabel('Temperature')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
