```python
# Dataset 1 - Employee Salary and Experience

import pandas as pd
import numpy as np

salary=[25000,30000,28000,35000,40000,32000,29000]
experience=[1,2,2,4,5,3,2]

print("Mean:",np.mean(salary))
print("Median:",np.median(salary))
print("varience:",np.var(salary))
print("standed deviation:",np.std(salary))
print("Correlation:",np.corrcoef(salary,experience))

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))
sns.scatterplot(x=experience, y=salary)

plt.title('Salary vs. Experience')
plt.xlabel('Experience')
plt.ylabel('Salary')
plt.grid(True)
plt.show()

plt.figure(figsize=(8, 6))
sns.histplot(salary, kde=True)

plt.title('Distribution of Salary')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
