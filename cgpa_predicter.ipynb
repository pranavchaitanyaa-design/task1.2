```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# Dataset
data = pd.DataFrame({
    "Marks": [650, 680, 700, 720, 750, 780],
    "CGPA": [7.0, 7.3, 7.5, 7.8, 8.1, 8.4]
})

# Train Model
X = data[["Marks"]]
y = data["CGPA"]

model = LinearRegression()
model.fit(X, y)

# Predict 7th Semester CGPA
sem7_marks = [[800]]
predicted_cgpa = model.predict(sem7_marks)

print("Predicted 7th Semester CGPA:", round(predicted_cgpa[0], 2))
import matplotlib.pyplot as plt

```
